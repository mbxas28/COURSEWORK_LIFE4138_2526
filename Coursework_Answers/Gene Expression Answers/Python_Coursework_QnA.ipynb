### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
import pandas as pd

life_expectancy = sns.load_dataset('healthexp')
````

````python
life_expectancy.head()
````

````python
#Finding total number of countries
num_countries = life_expectancy['Country'].nunique()
print("Number of unique countries:", num_countries)
````

````python

#Finding the country with the highest life expectancy and the year
max_life = life_expectancy.loc[life_expectancy['Life_Expectancy'].idxmax(), ['Country', 'Year', 'Life_Expectancy']]
print("Country and year with highest life expectancy:")
print(max_life)
````

````python

#Finding the year with the highest expenditure
yearly_expenditure = life_expectancy.groupby('Year')['Spending_USD'].sum()
max_expenditure_year = yearly_expenditure.idxmax()
print("Year with highest total expenditure:", max_expenditure_year)
````

### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline 

diamonds.head()

diamonds.plot.scatter(x='carat', y='price', alpha=1, color='Orange')
plt.title('Carat vs. Price')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()
````

````python
diamonds['price'].plot.hist(bins=20, color='skyblue', edgecolor='black')
plt.title('Prices of Diamonds')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
````

````python
#creating a boxplot
sns.boxplot(x="cut", y="price", data=diamonds)

#adding titles and labels
plt.title("Diamond Price by Cut")
plt.xlabel("Cut")
plt.ylabel("Price")

plt.show()
````
